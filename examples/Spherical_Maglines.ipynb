```python
Brtp = np.load('Brtp_pfss.npy')      # user's data
Nr,Nt,Np = Brtp.shape[1:]
r = np.linspace(1,2.5,Nr)           # rrange of user's data
t = np.linspace(np.pi,0,Nt)         # theta range of user's data
p = np.linspace(0,2*np.pi,Np)       # phi range of user's data
R,T,P = np.meshgrid(r,t,p,indexing='ij')
RTP = np.stack([R,T,P], axis=-1)
sd = spherical_data(
    grid=[r,t,p],
    fnames=['Br','Bt','Bp'],
    fields=Brtp
    )
```

```python
r_random = np.random.uniform(1,2.5,10)
t_random = np.random.uniform(np.pi,0,10)
p_random = np.random.uniform(0,2*np.pi,10)
RTP_random = np.stack([r_random,t_random,p_random],axis=-1) # user's  points coordinates (spherical coordinates)
maglines = sd.get_magline(RTP_random) # sd.get_magline(user's  points coordinates)
```

```python
fig = sd.show_maglines(maglines, c='magfield', lw=5, width=800, height=800,frame=False)
```
